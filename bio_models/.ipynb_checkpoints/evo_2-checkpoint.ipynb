```python
import os
os.environ["USE_TE"] = "0"

from evo2 import Evo2

# Load 7B model on L40S GPU
model = Evo2('evo2_7b')

output = model.generate(
    prompt="ATGCGATCGATCGATCG",
    max_new_tokens=100,
    temperature=0.7
)
print("Generated DNA Output:\n", output)
```
